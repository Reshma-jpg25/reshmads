```python

def find_median_sorted_arrays(nums1: list[int],
                              nums2: list[int]) -> float:
    """
    Find the median of two sorted arrays in O(log(min(m, n))) time.

    Args:
        nums1 (list[int]): First sorted array.
        nums2 (list[int]): Second sorted array.

    Returns:
        float: Median of the combined sorted arrays.
    """

    # Always perform binary search on the smaller array.
    if len(nums1) > len(nums2):
        nums1, nums2 = nums2, nums1

    m = len(nums1)
    n = len(nums2)

    left = 0
    right = m

    while left <= right:
        # Partition the first array.
        partition1 = (left + right) // 2

        # Partition the second array so the left half
        # contains half of all elements.
        partition2 = (m + n + 1) // 2 - partition1

        # Values immediately around the partitions.
        max_left1 = (
            float("-inf") if partition1 == 0
            else nums1[partition1 - 1]
        )

        min_right1 = (
            float("inf") if partition1 == m
            else nums1[partition1]
        )

        max_left2 = (
            float("-inf") if partition2 == 0
            else nums2[partition2 - 1]
        )

        min_right2 = (
            float("inf") if partition2 == n
            else nums2[partition2]
        )

        # Check whether both partitions are correctly placed.
        if max_left1 <= min_right2 and max_left2 <= min_right1:

            # Odd number of total elements.
            if (m + n) % 2 == 1:
                return float(max(max_left1, max_left2))

            # Even number of total elements.
            left_max = max(max_left1, max_left2)
            right_min = min(min_right1, min_right2)

            return (left_max + right_min) / 2.0

        # The first partition is too far to the right.
        elif max_left1 > min_right2:
            right = partition1 - 1

        # The first partition is too far to the left.
        else:
            left = partition1 + 1

    # Unreachable for valid sorted arrays.
    raise ValueError("Input arrays must be sorted.")


# Test cases
if __name__ == "__main__":
    test_cases = [
        ([1, 3], [2]),
        ([1, 2], [3, 4]),
        ([0, 0], [0, 0]),
        ([], [1]),
        ([2], []),
        ([1, 3, 5], [2, 4, 6])
    ]

    for nums1, nums2 in test_cases:
        median = find_median_sorted_arrays(nums1, nums2)

        print("nums1:", nums1)
        print("nums2:", nums2)
        print(f"Median: {median:.5f}")
        print("-" * 40)
```
