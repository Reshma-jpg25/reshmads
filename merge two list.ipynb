```python
# Merge Two Sorted Linked Lists
# Python - Jupyter Notebook / GitHub Version

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Add a node at the end
    def append(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next:
            current = current.next

        current.next = new_node

    # Display the linked list
    def display(self):
        current = self.head
        values = []

        while current:
            values.append(str(current.data))
            current = current.next

        print(" -> ".join(values) if values else "Empty List")


def merge_sorted_lists(list1, list2):
    # Create a dummy node
    dummy = Node(0)
    current = dummy

    # Compare nodes from both lists
    while list1 and list2:

        if list1.data <= list2.data:
            current.next = list1
            list1 = list1.next

        else:
            current.next = list2
            list2 = list2.next

        current = current.next

    # Attach remaining nodes
    if list1:
        current.next = list1
    else:
        current.next = list2

    return dummy.next


def display_nodes(head):
    values = []

    while head:
        values.append(str(head.data))
        head = head.next

    return " -> ".join(values) if values else "Empty List"


# --------------------------------------------------
# MAIN PROGRAM
# --------------------------------------------------

# First sorted linked list
list1 = LinkedList()

for value in [1, 2, 4]:
    list1.append(value)

# Second sorted linked list
list2 = LinkedList()

for value in [1, 3, 4]:
    list2.append(value)


# Display input lists
print("List 1:")
list1.display()

print("\nList 2:")
list2.display()


# Merge the two sorted lists
merged_head = merge_sorted_lists(list1.head, list2.head)


# Display merged list
print("\nMerged Sorted List:")
print(display_nodes(merged_head))
```
