```python

import heapq
from typing import Optional


class ListNode:
    """Node of a singly linked list."""

    def __init__(
        self,
        val: int = 0,
        next: Optional["ListNode"] = None
    ):
        self.val = val
        self.next = next


def merge_k_lists(
    lists: list[Optional[ListNode]]
) -> Optional[ListNode]:
    """
    Merge k sorted linked lists into one sorted linked list.

    Args:
        lists: A list containing the heads of sorted linked lists.

    Returns:
        The head of the merged sorted linked list.
    """

    min_heap = []

    # Add the first node of each non-empty list.
    for index, node in enumerate(lists):
        if node is not None:
            # The index breaks ties when node values are equal.
            heapq.heappush(min_heap, (node.val, index, node))

    # Dummy node simplifies construction of the result.
    dummy = ListNode()
    current = dummy

    while min_heap:
        # Remove the node with the smallest value.
        value, index, node = heapq.heappop(min_heap)

        # Append it to the merged linked list.
        current.next = node
        current = current.next

        # Add the next node from the same list, if available.
        if node.next is not None:
            heapq.heappush(
                min_heap,
                (node.next.val, index, node.next)
            )

    # Ensure the merged list ends correctly.
    current.next = None

    return dummy.next


def build_linked_list(values: list[int]) -> Optional[ListNode]:
    """Convert a Python list into a linked list."""

    dummy = ListNode()
    current = dummy

    for value in values:
        current.next = ListNode(value)
        current = current.next

    return dummy.next


def linked_list_to_list(
    head: Optional[ListNode]
) -> list[int]:
    """Convert a linked list into a Python list for display."""

    result = []

    while head is not None:
        result.append(head.val)
        head = head.next

    return result


# Test cases
if __name__ == "__main__":
    test_cases = [
        [[1, 4, 5], [1, 3, 4], [2, 6]],
        [],
        [[]],
        [[1]],
        [[-3, -1, 4], [-2, 0, 5]]
    ]

    for test_case in test_cases:
        linked_lists = [
            build_linked_list(values)
            for values in test_case
        ]

        merged_head = merge_k_lists(linked_lists)
        output = linked_list_to_list(merged_head)

        print("Input:", test_case)
        print("Output:", output)
        print("-" * 45)

```
