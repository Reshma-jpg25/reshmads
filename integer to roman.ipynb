```python
def int_to_roman(num: int) -> str:
    """
    Convert an integer into a Roman numeral.

    Args:
        num (int): An integer between 1 and 3999.

    Returns:
        str: The corresponding Roman numeral.
    """

    # Roman numeral values arranged from largest to smallest.
    roman_values = [
        (1000, "M"),
        (900, "CM"),
        (500, "D"),
        (400, "CD"),
        (100, "C"),
        (90, "XC"),
        (50, "L"),
        (40, "XL"),
        (10, "X"),
        (9, "IX"),
        (5, "V"),
        (4, "IV"),
        (1, "I")
    ]

    result = []

    # Process each Roman numeral value.
    for value, symbol in roman_values:

        # Add the symbol while its value fits into num.
        while num >= value:
            result.append(symbol)
            num -= value

    # Join all symbols into a single string.
    return "".join(result)


# Test cases
if __name__ == "__main__":
    test_cases = [3749, 58, 1994, 4, 9, 40, 400, 3999]

    for num in test_cases:
        roman = int_to_roman(num)
        print(f"Number: {num} -> Roman numeral: {roman}")
```
