```python
def is_match(s: str, p: str) -> bool:
    """
    Check whether the entire string matches the pattern.

    '.' matches any single character.
    '*' matches zero or more occurrences of the preceding element.
    """

    m = len(s)
    n = len(p)

    # dp[i][j] indicates whether s[:i] matches p[:j].
    dp = [[False] * (n + 1) for _ in range(m + 1)]

    # Empty string matches an empty pattern.
    dp[0][0] = True

    # Handle patterns such as a*, a*b*, and a*b*c*.
    for j in range(2, n + 1):
        if p[j - 1] == '*':
            dp[0][j] = dp[0][j - 2]

    # Fill the DP table.
    for i in range(1, m + 1):
        for j in range(1, n + 1):

            # Case 1: Characters match or pattern has '.'.
            if p[j - 1] == '.' or p[j - 1] == s[i - 1]:
                dp[i][j] = dp[i - 1][j - 1]

            # Case 2: Pattern has '*'.
            elif p[j - 1] == '*':
                # Treat the preceding element as occurring zero times.
                dp[i][j] = dp[i][j - 2]

                # Treat it as occurring one or more times.
                if p[j - 2] == '.' or p[j - 2] == s[i - 1]:
                    dp[i][j] = dp[i][j] or dp[i - 1][j]

    return dp[m][n]


# Test cases
if __name__ == "__main__":
    test_cases = [
        ("aa", "a", False),
        ("aa", "a*", True),
        ("ab", ".*", True),
        ("aab", "c*a*b", True),
        ("mississippi", "mis*is*p*.", False),
        ("aaa", "a*a", True),
        ("ab", ".*c", False),
        ("a", "ab*", True)
    ]

    for s, p, expected in test_cases:
        result = is_match(s, p)

        print(f"String:   {s}")
        print(f"Pattern:  {p}")
        print(f"Result:   {result}")
        print(f"Expected: {expected}")
        print(f"Passed:   {result == expected}")
        print("-" * 35)


```
