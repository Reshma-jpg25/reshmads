```python

def generate_parenthesis(n: int) -> list[str]:
    """
    Generate all combinations of well-formed parentheses.

    Args:
        n (int): Number of pairs of parentheses.

    Returns:
        list[str]: All valid combinations.
    """

    result = []
    current = []

    def backtrack(open_count: int, close_count: int) -> None:
        # If the string contains 2*n characters, it is complete.
        if len(current) == 2 * n:
            result.append("".join(current))
            return

        # Add '(' if we have not used all opening parentheses.
        if open_count < n:
            current.append("(")
            backtrack(open_count + 1, close_count)
            current.pop()  # Backtrack

        # Add ')' only when it cannot make the string invalid.
        if close_count < open_count:
            current.append(")")
            backtrack(open_count, close_count + 1)
            current.pop()  # Backtrack

    backtrack(0, 0)
    return result


# Test cases
if __name__ == "__main__":
    test_cases = [1, 2, 3, 4]

    for n in test_cases:
        combinations = generate_parenthesis(n)

        print(f"Input: n = {n}")
        print("Output:", combinations)
        print("Total combinations:", len(combinations))
        print("-" * 60)

        
```
