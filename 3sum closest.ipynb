```python
def three_sum_closest(nums: list[int], target: int) -> int:
    """
    Find the sum of three integers closest to the target.

    Args:
        nums (list[int]): Input integer array.
        target (int): Target sum.

    Returns:
        int: The closest possible sum of three elements.
    """

    # Sort the array to use the two-pointer technique.
    nums.sort()

    # Initialize with the sum of the first three elements.
    closest_sum = nums[0] + nums[1] + nums[2]

    # Fix the first element.
    for i in range(len(nums) - 2):

        left = i + 1
        right = len(nums) - 1

        # Search for the best pair for the fixed element.
        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]

            # If the sum exactly matches the target, return it.
            if current_sum == target:
                return current_sum

            # Update the closest sum if this one is better.
            if abs(current_sum - target) < abs(closest_sum - target):
                closest_sum = current_sum

            # Move pointers according to the current sum.
            if current_sum < target:
                left += 1
            else:
                right -= 1

    return closest_sum


# Test cases
if __name__ == "__main__":
    test_cases = [
        ([-1, 2, 1, -4], 1),
        ([0, 0, 0], 1),
        ([1, 1, 1, 0], 100),
        ([1, 2, 4, 8, 16], 15),
        ([-5, -2, -1, 4, 9], 2)
    ]

    for nums, target in test_cases:
        result = three_sum_closest(nums.copy(), target)

        print("Input array:", nums)
        print("Target:", target)
        print("Closest sum:", result)
        print("-" * 40)
```
