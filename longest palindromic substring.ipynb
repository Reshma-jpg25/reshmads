```python
def longest_palindrome(s: str) -> str:
    """
    Find the longest palindromic substring in a given string.

    A palindrome is a string that reads the same forward and backward.

    Args:
        s (str): Input string containing digits and English letters.

    Returns:
        str: The longest palindromic substring.
    """

    if not s:
        return ""

    start = 0
    max_length = 1

    def expand_around_center(left: int, right: int) -> None:
        """
        Expand outward while the characters match.
        Update the longest palindrome found so far.
        """
        nonlocal start, max_length

        while (
            left >= 0
            and right < len(s)
            and s[left] == s[right]
        ):
            current_length = right - left + 1

            if current_length > max_length:
                start = left
                max_length = current_length

            left -= 1
            right += 1

    for i in range(len(s)):
        # Case 1: Odd-length palindrome (example: "racecar")
        expand_around_center(i, i)

        # Case 2: Even-length palindrome (example: "abba")
        expand_around_center(i, i + 1)

    return s[start:start + max_length]


# Test the program
if __name__ == "__main__":
    test_cases = [
        "babad",
        "cbbd",
        "a",
        "racecar",
        "abba",
        "abcde",
        "aaaa"
    ]

    for test_string in test_cases:
        result = longest_palindrome(test_string)
        print(f"Input:  {test_string}")
        print(f"Output: {result}")
        print("-" * 30)
```
