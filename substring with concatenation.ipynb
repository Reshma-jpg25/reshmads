```python

from collections import Counter, defaultdict


def find_substring(s: str, words: list[str]) -> list[int]:
    """
    Find starting indices of substrings containing every word
    exactly as many times as it appears in words.

    Args:
        s (str): The input string.
        words (list[str]): Words of equal length.

    Returns:
        list[int]: Starting indices of valid concatenations.
    """

    if not s or not words:
        return []

    word_length = len(words[0])
    word_count = len(words)
    total_length = word_length * word_count

    # If the required substring is longer than s, no match exists.
    if total_length > len(s):
        return []

    # Required frequency of each word.
    word_frequency = Counter(words)
    result = []

    # Each offset represents a different alignment of word-sized chunks.
    for offset in range(word_length):
        left = offset
        seen = defaultdict(int)
        count = 0

        # Move right in chunks of word_length.
        for right in range(
            offset, len(s) - word_length + 1, word_length
        ):
            word = s[right:right + word_length]

            # If the word is not required, reset the window.
            if word not in word_frequency:
                seen.clear()
                count = 0
                left = right + word_length
                continue

            seen[word] += 1
            count += 1

            # Remove excess occurrences from the left.
            while seen[word] > word_frequency[word]:
                left_word = s[left:left + word_length]
                seen[left_word] -= 1
                count -= 1
                left += word_length

            # A valid window contains all required words.
            if count == word_count:
                result.append(left)

                # Move forward to search for overlapping matches.
                left_word = s[left:left + word_length]
                seen[left_word] -= 1
                count -= 1
                left += word_length

    return sorted(result)


# Test cases
if __name__ == "__main__":
    test_cases = [
        ("barfoothefoobarman", ["foo", "bar"]),
        ("wordgoodgoodgoodbestword",
         ["word", "good", "best", "word"]),
        ("barfoofoobarthefoobarman", ["bar", "foo", "the"]),
        ("aaaaaa", ["aa", "aa"]),
        ("foobar", ["foo", "bar"])
    ]

    for s, words in test_cases:
        result = find_substring(s, words)

        print("String:", s)
        print("Words:", words)
        print("Starting indices:", result)
        print("-" * 50)




```
